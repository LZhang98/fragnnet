```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem import rdmolops
from rdkit.Chem import Descriptors
from rdkit.Chem import AllChem
from copy import deepcopy
import pandas as pd

def draw_highlighted_atoms(mol, atom_ids):
    
    local_mol = deepcopy(mol)  # Create a deep copy of the molecule to avoid modifying the original
    for i, atom in enumerate(local_mol.GetAtoms()):
        # For each atom, set the property "molAtomMapNumber" to a custom number, let's say, the index of the atom in the molecule
        atom.SetProp("molAtomMapNumber", str(atom.GetIdx()))
    
    # Create a list of atom indices to highlight
    highlight_atoms = set(atom_ids)

    # Generate image with highlighted atoms
    img = Draw.MolToImage(local_mol, highlightAtoms=highlight_atoms, size=(300, 300))

    return img
```

```python
smarts_fp = "../../data/smarts/fraggnn_d3_mi_ma/smarts_train.csv"
smarts_df = pd.read_csv(smarts_fp)
```

```python
print("Count of formulas per prec_type:")
print(smarts_df.groupby('prec_type')['formula'].count())

print("\nCount of SMILES per prec_type:")
print(smarts_df.groupby('prec_type')['smiles'].count())
```

```python
# Count the number of unique formulas for each prec_type
unique_formula_counts = smarts_df.groupby('prec_type')['formula'].nunique()
print("Number of unique formulas for each prec_type:")
print(unique_formula_counts)

# Find overlapping formulas between different prec_types
overlapping_formulas = smarts_df.groupby('prec_type')['formula'].apply(set).agg(lambda x: set.intersection(*x))
print("\nNumber of overlapping formulas between prec_types:")
print(len(overlapping_formulas))
```

```python
# Count the number of unique SMILES for each prec_type
unique_smiles_counts = smarts_df.groupby('prec_type')['smiles'].nunique()
print("Number of unique SMILES for each prec_type:")
print(unique_smiles_counts)

# Find overlapping SMILES between different prec_types
overlapping_smiles = smarts_df.groupby('prec_type')['smiles'].apply(set).agg(lambda x: set.intersection(*x))
print("\nNumber of overlapping SMILES between prec_types:")
print(len(overlapping_smiles))
```
