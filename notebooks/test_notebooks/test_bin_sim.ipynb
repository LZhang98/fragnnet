```python
import torch as th
from frag_gnn.utils.spec_utils import batched_bin_func, calculate_sparse_cosine_similarity
```

```python
peaks1 = [(455.3, 999.0), (165.09, 952.95), (303.21, 345.55), (150.07, 75.92), (456.29, 62.84)]
mzs1, ints1 = [], []
for peak in peaks1:
    mzs1.append(peak[0])
    ints1.append(peak[1])
mzs1 = th.tensor(mzs1)
ints1 = th.tensor(ints1)
idxs1 = th.zeros_like(mzs1,dtype=th.long)
```

```python
peaks2 = [(455.291, 999.0), (456.2939, 233.76937662), (165.0902, 60.12750231), (457.2963, 33.973702290000006), (303.2061, 22.829467680000004)]
mzs2, ints2 = [], []
for peak in peaks2:
    mzs2.append(peak[0])
    ints2.append(peak[1])
mzs2 = th.tensor(mzs2)
ints2 = th.tensor(ints2)
idxs2 = th.zeros_like(mzs2,dtype=th.long)
```

```python
mz_max = 1005
mz_bin_res = 0.1
bins1_lr = batched_bin_func(
    mzs1,
    ints1,
    idxs1,
    mz_max,
    mz_bin_res,
    agg="sum",
    sparse=True
)
bins2_lr = batched_bin_func(
    mzs2,
    ints2,
    idxs2,
    mz_max,
    mz_bin_res,
    agg="sum",
    sparse=True
)
cos_sim_lr = calculate_sparse_cosine_similarity(
    mzs1,
    ints1,
    idxs1,
    mzs2,
    ints2,
    idxs2,
    mz_max,
    mz_bin_res,
    sum_ints=True
)
print(bins1_lr[0])
print(bins2_lr[0])
print(cos_sim_lr)
```

```python
mz_max = 1005
mz_bin_res = 0.01
bins1_hr = batched_bin_func(
    mzs1,
    ints1,
    idxs1,
    mz_max,
    mz_bin_res,
    agg="sum",
    sparse=True
)
bins2_hr = batched_bin_func(
    mzs2,
    ints2,
    idxs2,
    mz_max,
    mz_bin_res,
    agg="sum",
    sparse=True
)
cos_sim_hr = calculate_sparse_cosine_similarity(
    mzs1,
    ints1,
    idxs1,
    mzs2,
    ints2,
    idxs2,
    mz_max,
    mz_bin_res,
    sum_ints=True
)
print(bins1_hr[0])
print(bins2_hr[0])
print(cos_sim_hr)
```
