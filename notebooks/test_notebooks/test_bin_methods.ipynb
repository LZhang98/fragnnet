```python
import torch as th
```

```python
mzs = th.tensor([0., 25.3, 50.5, 50.4, 99.9, 100., 100.1])
mz_bin_res = 0.5
mz_max = 100.
bins = th.arange(mz_bin_res, mz_bin_res + mz_max, step=mz_bin_res)
```

```python
print(bins)
```

```python
idxs_1 = th.bucketize(mzs, bins, right=True)
print(idxs_1)
idxs_2 = th.searchsorted(bins, mzs, side="right")
print(idxs_2)
```

```python
idxs_1 = th.bucketize(mzs, bins, right=False)
print(idxs_1)
idxs_2 = th.searchsorted(bins, mzs, side="left")
print(idxs_2)
```
