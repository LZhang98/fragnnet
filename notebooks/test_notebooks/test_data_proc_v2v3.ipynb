## Differences (v2 vs v3)

mona23_df.csv, nist20_hr_df.csv -> no differences

spec_df, ann_df, no_mol_diff -> all NIST dset_spec_id are now type "str" in v3 (instead of type "int" in v2)

ann_df -> index is reset in v3

mol_df -> "inchi" column now holds actual InChI string (as opposed to InChIkey-14)

One additional difference with the spec_df (different inst_type) 
spec_id = 1094040
mol_id = 3945
inst_type = "Other" (v2) vs "FT" (v3)

```python
# v3 stuff
proc_dp_3 = "../data/proc/nist20"
spec_df_3 = pd.read_pickle(f"{proc_dp_3}/spec_df.pkl")
mol_df_3 = pd.read_pickle(f"{proc_dp_3}/mol_df.pkl")
ann_df_3 = pd.read_pickle(f"{proc_dp_3}/ann_df.pkl")
no_mol_df_3 = pd.read_pickle(f"{proc_dp_3}/no_mol_df.pkl")
diff_formula_df_3 = pd.read_pickle(f"{proc_dp_3}/diff_formula_df.pkl")
diff_inchikey_df_3 = pd.read_pickle(f"{proc_dp_3}/diff_inchikey_df.pkl")
diff_mass_df_3 = pd.read_pickle(f"{proc_dp_3}/diff_mass_df.pkl")
```

```python
def diff_df_mask(df1, df2):
    return (df1 != df2) & ~(df1.isna() | df2.isna())
```

```python
spec_diff = diff_df_mask(spec_df_2, spec_df_3)
print(spec_diff.sum())
```
