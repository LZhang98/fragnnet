```python
def sample_one_row_per_mol_id(df, n=10, random_state=69):
    mol_ids = df["mol_id"].drop_duplicates()
    if len(mol_ids) < n:
        raise ValueError(f"Expected at least {n} unique mol_ids, found {len(mol_ids)}")

    selected_mol_ids = mol_ids.sample(n=n, random_state=random_state)
    return (
        df[df["mol_id"].isin(selected_mol_ids)]
        .groupby("mol_id", group_keys=False)
        .sample(n=1, random_state=random_state)
        .reset_index(drop=True)
    )

sampled_df = sample_one_row_per_mol_id(df)
sampled_df
```
