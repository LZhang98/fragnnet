```python
import pandas as pd

msgym_fp = '../data/raw/msgym/MassSpecGym_with_test.tsv'
msgym_df = pd.read_csv(msgym_fp,sep="\t")
```

```python
msgym_df.columns
```

```python
msgym_df['has_ce'] = ~pd.isnull(msgym_df['collision_energy'])
```

```python
msgym_df.head()
```

Stats

```python
msgym_df.groupby(['adduct']).agg({'smiles': 'nunique'})
```

```python
msgym_df.groupby(['adduct','instrument_type', 'has_ce','fold','simulation_challenge']).agg({'identifier': 'nunique', 'smiles': 'nunique'})
```

Check CE Range

```python
msgym_df[msgym_df['has_ce']].groupby(['adduct','instrument_type','fold','simulation_challenge']).agg({'collision_energy': ['min','max','mean']})
```
