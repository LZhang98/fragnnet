```python
import torch
import pandas as pd
from IPython.display import display
```

```python
graff_formula_stats_df = pd.read_csv("../figs/formula/v3-1_feb25_scaffold_graff/formula_stats_df.csv")
nograff_formula_stats_df = pd.read_csv("../figs/formula/v3-1_feb25_scaffold_nograff/formula_stats_df.csv")
formula_stats_df = pd.concat([graff_formula_stats_df, nograff_formula_stats_df])
display(formula_stats_df)
```

```python
# average by seed
# formula_stats_df = formula_stats_df.drop(columns=["seed","split_type"]).groupby(["model_type"]).mean().reset_index()
formula_stats_df = formula_stats_df[["model_type","both_peak_coverage_0.0_mean","both_w_peak_coverage_0.0_mean","mode_peak_coverage_mean","mode_w_peak_coverage_mean"]].groupby(["model_type"]).agg(["mean","std"]).reset_index()
display(formula_stats_df)
```

```python
model_to_label = {
    "fraggnn_d4_long": "FraGNNet-D4",
    "graff3_long": "GrAFF",
    "iceberg_inten3_100_long": "ICEBERG"
}

column_to_label = {
    "model_type": "Model",
    "both_peak_coverage_0.0_mean": "Any Peak Coverage",
    "both_w_peak_coverage_0.0_mean": "Any Peak Weighted Coverage",
    "mode_peak_coverage_mean": "Mode Peak Coverage",
    "mode_w_peak_coverage_mean": "Mode Weighted Peak Coverage",
}

formula_stats_df["model_type"] = formula_stats_df["model_type"].map(model_to_label)
formula_stats_df = formula_stats_df[list(column_to_label.keys())]
formula_stats_df = formula_stats_df.rename(columns=column_to_label)
display(formula_stats_df)
```

```python
metrics = list(column_to_label.values())[1:]

for metric in metrics:
    formula_stats_df[metric] = formula_stats_df[metric].apply(
        lambda row: "-" if pd.isna(row['mean']) or pd.isna(row['std'])
                    else f"{row['mean']:.2f} $\pm$ {row['std']:.2f}", 
        axis=1
    )

formula_stats_df_str = formula_stats_df.to_latex(index=False, escape=False)

display(formula_stats_df_str)
```
