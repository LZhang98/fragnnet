```python
import pandas as pd

raw_fp = "/home/luke/fragnnet/data/raw/plastchem/Full_database.csv"
out_fp = "/home/luke/fragnnet/data/df/plastchem_df.csv"

raw = pd.read_csv(raw_fp, low_memory=False)
df = raw[["Identifiers.plastchem_ID", "Properties.isomeric_smiles"]].rename(
    columns={"Identifiers.plastchem_ID": "id", "Properties.isomeric_smiles": "smiles"}
)

# A row can only be predicted with both an id and a SMILES. Rows without an id are the empty
# rows at the end of the file; rows with an id but no SMILES are compounds PubChem holds no
# structure for (the file has no InChI for them either).
n_total = len(df)
no_id = df["id"].isna()
no_smiles = df["id"].notna() & df["smiles"].isna()
df = df[~no_id & ~no_smiles].reset_index(drop=True)
df["id"] = df["id"].astype(int)
assert df["id"].is_unique

df.to_csv(out_fp, index=False)

print(f"rows in database:                       {n_total}")
print(f"written (id and SMILES):                {len(df)}  ({df['id'].nunique()} ids, {df['smiles'].nunique()} unique SMILES)")
print(f"dropped, no id (empty trailing rows):   {no_id.sum()}")
print(f"dropped, id but no SMILES:              {no_smiles.sum()}")
assert len(df) + no_id.sum() + no_smiles.sum() == n_total
print(f"saved to {out_fp}")
df.head()
```
