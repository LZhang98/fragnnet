```python
import numpy as np
import h5py
data_fp = "../../data/mces-data/mces_pairs_max10.h5"

with h5py.File(data_fp, 'r') as f:
    # Print the keys to see what's in the file
    print("Keys in the HDF5 file:")
    print(list(f.keys()))
 
    # Read the data
    mces = f['mces_pairs'][:]
    #smiles = f['smiles'][:]
    #computation_indices = f['computation_indices'][:]
    #print(mces.shape, smiles.shape, computation_indices.shape)
    #print("computation_indices:", computation_indices)
    #print("First 5 MCES values:", mces[:5])
    #print("First 5 SMILES strings:", smiles[:5])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.hist(mces, bins=50, alpha=0.7, edgecolor='black')
plt.xlabel('MCES Values')
plt.ylabel('Frequency')
plt.title('Distribution of MCES Values')
plt.grid(True, alpha=0.3)
plt.show()

# Print some statistics
print(f"MCES statistics:")
print(f"Mean: {np.mean(mces):.2f}")
print(f"Median: {np.median(mces):.2f}")
print(f"Min: {np.min(mces):.2f}")
print(f"Max: {np.max(mces):.2f}")
print(f"Standard deviation: {np.std(mces):.2f}")
```

```python
# Count rows with MCES values less than different thresholds
thresholds = [5, 10, 20]

for threshold in thresholds:
    count = np.sum(mces < threshold)
    percentage = (count / len(mces)) * 100
    print(f"MCES < {threshold}: {count} rows ({percentage:.2f}%)")

# Total number of rows
print(f"\nTotal rows: {len(mces)}")
```
