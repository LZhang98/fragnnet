```python
train_ids = pd.read_csv("../data/split/massspecgym/train_ids.csv")
val_ids = pd.read_csv("../data/split/massspecgym/val_ids.csv")
test_ids = pd.read_csv("../data/split/massspecgym/test_ids.csv")
```

```python
mol_ids_d3 = [int(os.path.basename(file).removesuffix(".pickle.bz2")) for file in os.listdir("../data/frag/massspecgym_d3/dags")]
mol_ids_d4 = [int(os.path.basename(file).removesuffix(".pickle.bz2")) for file in os.listdir("../data/frag/massspecgym_d4/dags")]
```

```python
print(len(mol_ids_d3))
print(len(mol_ids_d4))
```

```python
set(mol_ids_d3) - set(mol_ids_d4)
```

```python
set(mol_ids_d4) - set(mol_ids_d3)
```

```python
train_ids_f = train_ids[train_ids["mol_id"].isin(mol_ids_d3)]
val_ids_f = val_ids[val_ids["mol_id"].isin(mol_ids_d3)]
test_ids_f = test_ids[test_ids["mol_id"].isin(mol_ids_d3)]
os.makedirs("../data/split/massspecgym_f",exist_ok=True)
train_ids_f.to_csv("../data/split/massspecgym_f/train_ids.csv",index=False)
val_ids_f.to_csv("../data/split/massspecgym_f/val_ids.csv",index=False)
test_ids_f.to_csv("../data/split/massspecgym_f/test_ids.csv",index=False)
```
