```python
meta_fp = '../../data/raw/spectraverse/metadata-can-rem-lowres-prec-matchms-dotcheck-noiserem-fold-metadata.csv.gz'
meta_df = pd.read_csv(meta_fp)
meta_df.head()
```

```python
# Find duplicated SMILES
for i in range(1, 10):
    fold_smiles = meta_df[meta_df['FOLD'] == i]['SMILES']
    other_folds_smiles = meta_df[meta_df['FOLD'] != i]['SMILES']
    print(i , len(set(fold_smiles) - set(other_folds_smiles)) - len(set(fold_smiles)))
```
