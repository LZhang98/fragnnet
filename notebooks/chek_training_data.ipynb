```python
import pandas as pd

nist20_dp = '../data/proc/nist20mona23'
mol_df = pd.read_pickle(nist20_dp + '/mol_df.pkl')
spec_df = pd.read_pickle(nist20_dp + '/spec_df.pkl')
```

```python
spec_df.columns
```

```python
print("all dset",spec_df['dset'].unique())
print("all inst_type",spec_df['inst_type'].unique())

nist20_qtof_df = spec_df[(spec_df['dset'] == 'nist20_hr')]# & (spec_df['inst_type'] == 'QTOF')]
nist20_qtof_df = nist20_qtof_df[nist20_qtof_df['ace'].notna()]
print(f">nist20 ft df shape {nist20_qtof_df.shape}")
print(nist20_qtof_df['ace'].mean())
print(nist20_qtof_df['ace'].std())
print(nist20_qtof_df['ace'].min())
print(nist20_qtof_df['ace'].max())

#print(metric, df.loc[:, metric].mean(),  df.loc[:, metric].std())
#agg_keys = ['prec_type']
agg_key = 'prec_type'
print(f"Aggregated by {agg_key}, ace \n", nist20_qtof_df.groupby(agg_key)['ace'].agg(['mean','std','count','min','max']))
agg_key = 'inst_type'
print(f"Aggregated by {agg_key}, ace \n", nist20_qtof_df.groupby(agg_key)['ace'].agg(['mean','std','count','min','max']))
```

```python
nist20_ft_df = spec_df[(spec_df['dset'] == 'nist20_hr') & (spec_df['inst_type'] == 'FT')]
nist20_ft_df = nist20_ft_df[nist20_ft_df['nce'].notna()]
print(f">nist20 ft df shape {nist20_ft_df.shape}")
print(nist20_ft_df['nce'].mean())
print(nist20_ft_df['nce'].std())
print(nist20_ft_df['nce'].min())
print(nist20_ft_df['nce'].max())

agg_key = 'prec_type'
print(f"Aggregated by {agg_key}, nce \n", nist20_ft_df.groupby(agg_key)['nce'].agg(['mean','std','count']))
```
